```python
from bisect import bisect_left

def index(a, x):
    i = bisect_left(a, x)
    if i != len(a) and a[i] == x:
        return i
    raise ValueError

data = [10, 20, 20, 30, 40]
target = 20

hasil = index(data, target)
print("Indeks yang ditemukan:", hasil)
```

```python
data = [10, 20, 20, 30, 40]
target = 20

i = bisect_left(data, target)

print("Data:", data)
print("Target:", target)
print("Indeks i:", i)
print("Panjang daftar:", len(data))
print("i != len(data):", i != len(data))

if i != len(data):
    print("data[i]:", data[i])
    print("data[i] == target:", data[i] == target)

print("Hasil fungsi:", index(data, target))
```

```python
data = [10, 20, 20, 30, 40]

for target in [10, 20, 25, 40, 50]:
    try:
        posisi = index(data, target)
        print(f"Nilai {target}: indeks {posisi}")
    except ValueError:
        print(f"Nilai {target}: tidak ditemukan")
```

```python
kasus_uji = [
    ("P01", [], 20, "ValueError"),
    ("P02", [20], 20, 0),
    ("P03", [20], 10, "ValueError"),
    ("P04", [10, 20, 20, 30, 40], 20, 1),
    ("P05", [10, 20, 20, 30, 40], 40, 4),
    ("P06", [10, 20, 20, 30, 40], 25, "ValueError"),
    ("P07", [10, 20, 20, 30, 40], 50, "ValueError"),
    ("P08", [-10, 0, 10], -10, 0),
]

for kode, data, target, harapan in kasus_uji:
    try:
        aktual = index(data, target)
    except ValueError:
        aktual = "ValueError"

    status = "SESUAI" if aktual == harapan else "TIDAK SESUAI"
    print(f"{kode} | target={target} | "
          f"harapan={harapan} | aktual={aktual} | {status}")
```

```python
import unittest

# Fungsi revisi
def cari_indeks_pertama(data, target):
    # Daftar harus terurut menaik.
    posisi = bisect_left(data, target)
    if posisi < len(data) and data[posisi] == target:
        return posisi
    raise ValueError(f"Nilai {target!r} tidak ditemukan")


class TestPencarian(unittest.TestCase):
    def test_awal_dan_revisi(self):
        kasus = [
            ([], 20, None),
            ([20], 20, 0),
            ([20], 10, None),
            ([10, 20, 20, 30, 40], 20, 1),
            ([10, 20, 20, 30, 40], 40, 4),
            ([10, 20, 20, 30, 40], 25, None),
            ([10, 20, 20, 30, 40], 50, None),
            ([-10, 0, 10], -10, 0),
        ]

        for fungsi in (index, cari_indeks_pertama):
            for data, target, harapan in kasus:
                with self.subTest(
                    fungsi=fungsi.__name__,
                    data=data,
                    target=target
                ):
                    if harapan is None:
                        with self.assertRaises(ValueError):
                            fungsi(data, target)
                    else:
                        self.assertEqual(
                            fungsi(data, target), harapan
                        )


suite = unittest.defaultTestLoader.loadTestsFromTestCase(
    TestPencarian
)
unittest.TextTestRunner(verbosity=2).run(suite)
```
